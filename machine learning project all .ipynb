```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
```

```python
target = "monthly_bill_egp"
X = customers_clean.drop(
    columns=["monthly_bill_egp", "customer_id", "churn"],
    errors="ignore"
)
y = customers_clean[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nFeature columns:")
print(X.columns.tolist())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Training rows:", len(y_train))
print("Testing rows:", len(y_test))
```

```python
numeric_features = X_train.select_dtypes(
    include=["int64", "float64", "number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

print("Numeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore",sparse_output=False))
])
```

```python
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])
```

```python
models={
    "Dummy":DummyRegressor(strategy="mean"),
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "Random Forest" :RandomForestRegressor(n_estimators=100,random_state=42,n_jobs=-1),
    "Gradient Boosting" : GradientBoostingRegressor(random_state=42)
}
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import(
    RandomForestClassifier,
    GradientBoostingClassifier,
    ExtraTreesClassifier
)
```

```python
result=[]
trained_model={}
for name ,model in models.items():
    pipeline=Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])
    pipeline.fit(X_train, y_train)
    predection=pipeline.predict(X_test)
    mae=mean_absolute_error(y_test,predection)
    rmse=np.sqrt(mean_squared_error(y_test,predection))
    r2=r2_score(y_test,predection)
    result.append({
        "Model":name,
        "MAE":mae,
        "RMSE": rmse,
        "R2":r2
    } )
    trained_model[name]=pipeline
result_df=pd.DataFrame(result)
result_df=result_df.sort_values(by="MAE",ascending=True)
display(result_df)
```

```python
target="churn"
X=customers_clean.drop(columns=["churn","customer_id"],errors="ignore")
y=customers_clean[target]
print(X.shape)
print(y.shape)
print(y.value_counts())
```

```python
X_train ,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
print("Training churn distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting churn distribution:")
print(y_test.value_counts(normalize=True))
```

```python
numeric_features = X_train.select_dtypes(
    include="number"
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])
```

```python
models={
    "Logistic Regression": LogisticRegression(max_iter=2000,random_state=42),
    "KNN":KNeighborsClassifier(n_neighbors=5),
    "SVM":SVC(probability = True,random_state=42),
    "Decision Tree" : DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100,random_state=42,n_jobs=-1),
    "Gradient Boosting" : GradientBoostingClassifier(random_state=42),
    "Extra Trees": ExtraTreesClassifier(n_estimators=100,random_state=42,n_jobs=-1)
}
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)
```

```python
result=[]
trained_models={}
for name ,model in models.items():
    pipeline=Pipeline([
        ("Preprocessor",preprocessor),
        ("model",model)
    ])
    pipeline.fit(X_train ,y_train)
    predictions=pipeline.predict(X_test)
    probabilities=pipeline.predict_proba(X_test)[:,1]
    result.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions, pos_label="Yes", zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, pos_label="Yes", zero_division=0
        ),
        "F1": f1_score(
            y_test, predictions, pos_label="Yes", zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            (y_test == "Yes").astype(int), probabilities
        )
    })
    trained_models[name]=pipeline
result_df=pd.DataFrame(result)
display(
    result_df.sort_values(
        by="Recall",
        ascending=False
    )
)
```

```python
best_model_name=result_df.iloc[0]["Model"]
best_model=trained_models[best_model_name]
predictions=best_model.predict(X_test)
cm=confusion_matrix(
    y_test,
    predictions,
    labels=["No","Yes"]
)
print("Selected model:", best_model_name)
print(cm)
```

```python
tn, fp, fn, tp = cm.ravel()

print("Correctly predicted non-churn:", tn)
print("False alarms:", fp)
print("Missed churn customers:", fn)
print("Correctly identified churn:", tp)
```

```python
from sklearn.model_selection import StratifiedKFold ,cross_validate
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

cv_results = []

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        (y_train == "Yes").astype(int),
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "Mean Recall": scores["test_recall"].mean(),
        "Std Recall": scores["test_recall"].std(),
        "Mean F1": scores["test_f1"].mean(),
        "Mean ROC-AUC": scores["test_roc_auc"].mean()
    })

cv_results_df = pd.DataFrame(cv_results)

display(
    cv_results_df.sort_values(
        by="Mean Recall",
        ascending=False
    )
)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

X_cluster_numeric = X_cluster.select_dtypes(
    include="number"
)

cluster_imputer = SimpleImputer(strategy="median")

X_cluster_imputed = cluster_imputer.fit_transform(
    X_cluster_numeric
)

cluster_scaler = StandardScaler()

X_cluster_scaled = cluster_scaler.fit_transform(
    X_cluster_imputed
)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(
    X_cluster_scaled
)

print("Cluster counts:")
print(pd.Series(kmeans_labels).value_counts().sort_index())

print(
    "Silhouette Score:",
    silhouette_score(X_cluster_scaled, kmeans_labels)
)
```

```python
from sklearn.cluster import AgglomerativeClustering

hac = AgglomerativeClustering(
    n_clusters=4,
    linkage="ward"
)

hac_labels = hac.fit_predict(
    X_cluster_scaled
)

print("HAC cluster counts:")
print(pd.Series(hac_labels).value_counts().sort_index())

print(
    "HAC Silhouette:",
    silhouette_score(X_cluster_scaled, hac_labels)
)
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=1.5,
    min_samples=10
)

dbscan_labels = dbscan.fit_predict(
    X_cluster_scaled
)

print("DBSCAN labels and counts:")
print(pd.Series(dbscan_labels).value_counts().sort_index())
```

```python
clustering_results = []

for name, labels in [
    ("KMeans", kmeans_labels),
    ("HAC", hac_labels),
    ("DBSCAN", dbscan_labels)
]:
    n_clusters = len(set(labels) - {-1})
    noise_count = int(np.sum(labels == -1))

    if n_clusters >= 2:
        score = silhouette_score(
            X_cluster_scaled,
            labels
        )
    else:
        score = np.nan

    clustering_results.append({
        "Algorithm": name,
        "Number of clusters": n_clusters,
        "Noise points": noise_count,
        "Silhouette": score
    })

clustering_results_df = pd.DataFrame(clustering_results)

display(clustering_results_df)
```

```python
non_noise_mask = dbscan_labels != -1
non_noise_labels = dbscan_labels[non_noise_mask]
non_noise_data = X_cluster_scaled[non_noise_mask]

if len(set(non_noise_labels)) >= 2:
    dbscan_silhouette = silhouette_score(
        non_noise_data,
        non_noise_labels
    )
    print("DBSCAN Silhouette (excluding noise):", dbscan_silhouette)
else:
    print("DBSCAN did not find enough non-noise clusters for silhouette.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.base import clone
from sklearn.metrics import recall_score, f1_score, roc_auc_score
```

```python
target = "churn"

drop_columns = [
    "customer_id",
    "churn",
    "cluster"
]

X_base = customers_for_segments.drop(
    columns=drop_columns,
    errors="ignore"
)

X_with_segment = customers_for_segments.drop(
    columns=["customer_id", "churn"],
    errors="ignore"
)

y_segment = (
    customers_for_segments["churn"].eq("Yes").astype(int)
)

X_train_base, X_test_base, y_train_seg, y_test_seg = train_test_split(
    X_base,
    y_segment,
    test_size=0.20,
    random_state=42,
    stratify=y_segment
)

X_train_segment, X_test_segment, _, _ = train_test_split(
    X_with_segment,
    y_segment,
    test_size=0.20,
    random_state=42,
    stratify=y_segment
)
```

```python
def make_classifier_pipeline(X_data):
    numeric_cols = X_data.select_dtypes(
        include="number"
    ).columns.tolist()

    categorical_cols = X_data.select_dtypes(
        include=["object", "category", "string"]
    ).columns.tolist()

    numeric_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ])

    prep = ColumnTransformer([
        ("numeric", numeric_pipe, numeric_cols),
        ("categorical", categorical_pipe, categorical_cols)
    ])

    return Pipeline([
        ("preprocessor", prep),
        ("model", LogisticRegression(
            max_iter=2000,
            random_state=42
        ))
    ])
```

```python
base_model = make_classifier_pipeline(X_train_base)

segment_model = make_classifier_pipeline(X_train_segment)

base_model.fit(X_train_base, y_train_seg)

segment_model.fit(X_train_segment, y_train_seg)

base_predictions = base_model.predict(X_test_base)
segment_predictions = segment_model.predict(X_test_segment)

base_probabilities = base_model.predict_proba(X_test_base)[:, 1]
segment_probabilities = segment_model.predict_proba(X_test_segment)[:, 1]
```

```python
comparison = pd.DataFrame([
    {
        "Model": "Without Segment",
        "Recall": recall_score(y_test_seg, base_predictions),
        "F1": f1_score(y_test_seg, base_predictions),
        "ROC-AUC": roc_auc_score(y_test_seg, base_probabilities)
    },
    {
        "Model": "With Segment",
        "Recall": recall_score(y_test_seg, segment_predictions),
        "F1": f1_score(y_test_seg, segment_predictions),
        "ROC-AUC": roc_auc_score(y_test_seg, segment_probabilities)
    }
])

display(comparison)
```
